# FUTURE_DS_02 – Customer Retention & Churn Analysis

```python
import pandas as pd
df=pd.read_csv('../data/customer_retention.csv',parse_dates=['Signup_Date'])
df.head()
```

```python
print('Overall churn:',df.Churned.mean())
print(df.groupby('Plan').Churned.mean().sort_values(ascending=False))
print(df.groupby('Channel').Churned.mean().sort_values(ascending=False))
```

```python
cohort=df.groupby('Signup_Month').agg(Customers=('Customer_ID','count'),Churn=('Churned','sum'))
cohort['Retention_Rate']=1-cohort.Churn/cohort.Customers
cohort.tail()
```

## Recommendations
Improve onboarding for high-risk plans, test annual-plan incentives, audit high-churn acquisition channels, and monitor retention cohorts over time.
